```python
# take an example of fibonacci sequence :-

import time 

def fib(n):
    if n==0 or n==1:
        return 1
    else:
       return fib(n-1) + fib(n-2)
```

```python
start = time.time()

print(fib(100))

print(time.time() - start)
```

<h3> Now we use memoization method for reducing time to get an output.</h3>

```python
def fib_using_mem(n,d):
    if n in d:
        return d[n]
    else: 
        d[n] = fib_using_mem(n-1,d) + fib_using_mem(n-2,d)
        return d[n]
```

```python
start_2 = time.time()

d = {0:1, 1:1}

print(fib_using_mem(100,d))
print(time.time() - start_2)
```
